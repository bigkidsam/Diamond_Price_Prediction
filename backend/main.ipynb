```python
diamond.columns = ['carat', 'cut', 'color', 'clarity', 'depth', 'table', 'price', 'x', 'y', 'z']

for col in ['cut', 'color', 'clarity']:
    print(f"--- {col} Value Counts ---")
    print(diamond[col].value_counts(), "\n")

diamond.describe()
```

```python
print("x <= 0:", (diamond['x'] <= 0).sum())
print("y <= 0:", (diamond['y'] <= 0).sum())
print("z <= 0:", (diamond['z'] <= 0).sum())

# Display Sucpicious Records 
diamond[(diamond['x'] <= 0) | 
        (diamond['y'] <= 0) | 
        (diamond['z'] <= 0)]
```

```python
diamond = diamond[
    (diamond['x'] > 0) &
    (diamond['y'] > 0) &
    (diamond['z'] > 0)
].copy()

print("Shape after removing invalid dimensions:", diamond.shape)
```

```python
X = diamond.drop('price',axis=1)
y = diamond['price']

print("Features")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

**TRAIN / TEST SPLIT**

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(


    X,
    y,
    test_size=0.20,
    random_state=42

)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Combine training features and target for EDA
train_data = X_train.copy()
train_data['price'] = y_train

print(train_data.shape)
train_data.head()
```

```python
plt.figure (figsize=(8,5))

sns.histplot(
    data= train_data,
    x='price',

    bins=50,

    kde=True
)

plt.title("Diamond Price Distribution - Training Data")
plt.xlabel('Price ')
plt.ylabel('Frequency')
plt.show()
```

```python

plt.figure(figsize=(8,5))


sns.scatterplot(
    data = train_data,
    x='carat',
    y='price',
    alpha=0.3
)

plt.title('Carat vs Price -Training Data')
plt.xlabel('Carat')
plt.ylabel(' Price')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=train_data,
    x='cut',
    y='price'

)


plt.title('Price by cut - Training Data')
plt.xlabel(' cut')
plt.ylabel('Price')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=train_data,
    x='color',
    y='price'
)

plt.title('Price by Color - Training Data')
plt.xlabel('Color')
plt.ylabel('Price')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=train_data,
    x='clarity',
    y='price'
)

plt.title('Price by Clarity - Training Data')
plt.xlabel('Clarity')
plt.ylabel('Price')
plt.xticks(rotation=30)
plt.show()
```

```python
numeric_features = [
    'carat',
    'depth',
    'table',
    'price',
    'x',
    'y',
    'z'
]

plt.figure(figsize=(10, 7))

sns.heatmap(
    train_data[numeric_features].corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Matrix - Training Data')
plt.show()
```
